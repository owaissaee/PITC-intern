## **Lists, Tuples, Dictionaries, Sets**

```python
list1 = [10, 30, 20, 40, 50]
print("Original List:", list1)

list1[2] = 25 #modify
print("Modified List:", list1)

list1.append(60) #append
print("List after appending 60:", list1)

list1.remove(30) #remove
print("List after removing 30:", list1)

list1.sort() #sort
print("Sorted List:", list1)

list1.reverse() #reverse
print("Reversed List:", list1)

features = ["height", "weight", "age", "blood_pressure"]
print(f"First feature: {features[0]}")
print(f"Last feature: {features[-1]}")
```

**Sets**

```python
my_list = [10, 20, 20, 30, 40]
my_tuple = (10, 20, 20, 30, 40)
my_set = {10, 20, 20, 30, 40}

print("List:", my_list)
print("Tuple:", my_tuple)
print("Set:", my_set)

#Append
print("\nMutability")
my_list.append(40)
#my_tuple.append(40)
my_set.add(40)

print("Updated List:", my_list)
print("Updated Set:", my_set)

#indexing
print("\nIndexing")
print("List[1]:", my_list[1])
print("Tuple[1]:", my_tuple[1])
#print(my_set[1])
```

## If-else, Loops

```python
# If-elif-else
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"

# Ternary operator
status = "Pass" if score >= 60 else "Fail"

# For loops
for i in range(5):           # 0,1,2,3,4
    print(i)

for i in range(2, 8, 2):    # 2,4,6
    print(i)

# While loops
count = 0
while count < 5:
    print(count)
    count += 1

# Break/Continue
for i in range(10):
    if i == 3:
        continue    # Skip 3
    if i == 7:
        break       # Stop at 7
    print(i)        # 0,1,2,4,5,6
```
